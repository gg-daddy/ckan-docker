## 2. Load Dataset



Load the dataset from Hugging Face: `uoft-cs/cifar10`

```python
from datasets import load_dataset

import pandas as pd



dataset = load_dataset("uoft-cs/cifar10")



# Display dataset info

print(dataset)

print(f"\nTrain samples: {len(dataset['train'])}")
```

## 3. Explore Data



View sample data from the dataset.

```python
# Class labels
CLASS_NAMES = ['airplane', 'automobile', 'bird', 'cat', 'deer', 
               'dog', 'frog', 'horse', 'ship', 'truck']

# View a sample image
sample = dataset["train"][0]
print(f"Label: {sample['label']} ({CLASS_NAMES[sample['label']]})")
print(f"Image shape: {sample['img'].size}")

from IPython.display import display
display(sample["img"])
```

## 4. Visualize Data



Create visualizations to understand the data distribution.

```python
# Display a grid of images by class
import matplotlib.pyplot as plt

CLASS_NAMES = ['airplane', 'automobile', 'bird', 'cat', 'deer', 
               'dog', 'frog', 'horse', 'ship', 'truck']

fig, axes = plt.subplots(2, 5, figsize=(14, 6))
for label, ax in enumerate(axes.flat):
    # Find first image of this class
    for sample in dataset["train"]:
        if sample["label"] == label:
            ax.imshow(sample["img"])
            ax.set_title(CLASS_NAMES[label])
            ax.axis('off')
            break
plt.suptitle('CIFAR-10 Class Examples')
plt.tight_layout()
plt.show()
```

## 5. Convert to Pandas



For further analysis, convert to a pandas DataFrame.

```python
# Convert to pandas DataFrame

df = dataset['train'].to_pandas()



print(f"DataFrame shape: {df.shape}")

print(f"\nColumns: {list(df.columns)}")

print(f"\nData types:")

print(df.dtypes)



df.head()
```
