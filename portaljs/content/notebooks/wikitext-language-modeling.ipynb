## 2. Load Dataset



Load the dataset from Hugging Face: `Salesforce/wikitext`

```python
from datasets import load_dataset

import pandas as pd



dataset = load_dataset("Salesforce/wikitext", "wikitext-2-v1")



# Display dataset info

print(dataset)

print(f"\nTrain samples: {len(dataset['train'])}")
```

## 3. Explore Data



View sample data from the dataset.

```python
# View sample text
sample = dataset["train"][10]
print("Sample text:")
print(sample["text"][:1000])
```

## 4. Visualize Data



Create visualizations to understand the data distribution.

```python
# Analyze vocabulary
import collections

all_text = " ".join([ex["text"] for ex in dataset["train"][:1000]])
words = all_text.lower().split()

word_freq = collections.Counter(words).most_common(20)
print("Top 20 words:")
for word, count in word_freq:
    print(f"  {word}: {count}")

print(f"\nTotal tokens: {len(words)}")
print(f"Unique tokens: {len(set(words))}")
```

## 5. Convert to Pandas



For further analysis, convert to a pandas DataFrame.

```python
# Convert to pandas DataFrame

df = dataset['train'].to_pandas()



print(f"DataFrame shape: {df.shape}")

print(f"\nColumns: {list(df.columns)}")

print(f"\nData types:")

print(df.dtypes)



df.head()
```
