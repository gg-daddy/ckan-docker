## 2. Load Dataset



Load the dataset from Hugging Face: `ylecun/mnist`

```python
from datasets import load_dataset

import pandas as pd



dataset = load_dataset("ylecun/mnist")



# Display dataset info

print(dataset)

print(f"\nTrain samples: {len(dataset['train'])}")
```

## 3. Explore Data



View sample data from the dataset.

```python
# View a sample digit
sample = dataset["train"][0]
print(f"Label: {sample['label']}")
print(f"Image shape: {sample['image'].size}")

# Display the image
from IPython.display import display
display(sample["image"])
```

## 4. Visualize Data



Create visualizations to understand the data distribution.

```python
# Display a grid of sample digits
import matplotlib.pyplot as plt

fig, axes = plt.subplots(2, 5, figsize=(12, 5))
for i, ax in enumerate(axes.flat):
    sample = dataset["train"][i]
    ax.imshow(sample["image"], cmap='gray')
    ax.set_title(f"Label: {sample['label']}")
    ax.axis('off')
plt.suptitle('Sample MNIST Digits')
plt.tight_layout()
plt.show()

# Label distribution
import collections
labels = [ex["label"] for ex in dataset["train"]]
print(f"Label distribution: {dict(sorted(collections.Counter(labels).items()))}}")
```

## 5. Convert to Pandas



For further analysis, convert to a pandas DataFrame.

```python
# Convert to pandas DataFrame

df = dataset['train'].to_pandas()



print(f"DataFrame shape: {df.shape}")

print(f"\nColumns: {list(df.columns)}")

print(f"\nData types:")

print(df.dtypes)



df.head()
```
