## 2. Load Dataset



Load the dataset from Hugging Face: `fancyzhx/ag_news`

```python
from datasets import load_dataset

import pandas as pd



dataset = load_dataset("fancyzhx/ag_news")



# Display dataset info

print(dataset)

print(f"\nTrain samples: {len(dataset['train'])}")
```

## 3. Explore Data



View sample data from the dataset.

```python
# Class labels
CLASS_NAMES = ['World', 'Sports', 'Business', 'Sci/Tech']

# View sample articles
for i in range(4):
    sample = dataset["train"][i]
    print(f"[{CLASS_NAMES[sample['label']]}] {sample['text'][:150]}...")
    print()
```

## 4. Visualize Data



Create visualizations to understand the data distribution.

```python
# Label distribution
import matplotlib.pyplot as plt
import collections

CLASS_NAMES = ['World', 'Sports', 'Business', 'Sci/Tech']

labels = [ex["label"] for ex in dataset["train"]]
label_counts = collections.Counter(labels)

plt.figure(figsize=(10, 5))
plt.bar(CLASS_NAMES, [label_counts[i] for i in range(4)], color=['#ff6b6b', '#4ecdc4', '#45b7d1', '#96c93d'])
plt.ylabel('Count')
plt.title('AG News Category Distribution')
for i, v in enumerate([label_counts[i] for i in range(4)]):
    plt.text(i, v + 200, str(v), ha='center')
plt.show()
```

## 5. Convert to Pandas



For further analysis, convert to a pandas DataFrame.

```python
# Convert to pandas DataFrame

df = dataset['train'].to_pandas()



print(f"DataFrame shape: {df.shape}")

print(f"\nColumns: {list(df.columns)}")

print(f"\nData types:")

print(df.dtypes)



df.head()
```
