```python
import pandas as pd
import requests
from IPython.display import display, Markdown
```

## 1. Fetch Dataset Metadata

First, let's retrieve the dataset information from the CKAN API.

```python
# Configuration
CKAN_BASE_URL = "https://your-ckan-domain.com"  # Update this to your CKAN instance URL
DATASET_ID = "example-dataset"

# Fetch dataset metadata
api_url = f"{CKAN_BASE_URL}/api/3/action/package_show"
response = requests.get(api_url, params={"id": DATASET_ID})

if response.status_code == 200:
    dataset = response.json()["result"]
    print(f"Dataset: {dataset['title']}")
    print(f"Description: {dataset.get('notes', 'No description')[:200]}...")
else:
    print(f"Error fetching dataset: {response.status_code}")
```

## 2. List Available Resources

View all downloadable files in this dataset.

```python
# List resources
resources = dataset.get("resources", [])

print(f"Found {len(resources)} resource(s):\n")
for i, res in enumerate(resources, 1):
    print(f"{i}. {res['name']}")
    print(f"   Format: {res.get('format', 'Unknown')}")
    print(f"   URL: {res['url']}")
    print()
```

## 3. Load CSV Data

Load the first CSV resource into a pandas DataFrame.

```python
# Find CSV resources
csv_resources = [r for r in resources if r.get('format', '').upper() == 'CSV']

if csv_resources:
    csv_url = csv_resources[0]['url']
    print(f"Loading: {csv_resources[0]['name']}")
    df = pd.read_csv(csv_url)
    print(f"\nShape: {df.shape[0]} rows, {df.shape[1]} columns")
    display(df.head())
else:
    print("No CSV resources found in this dataset.")
```
